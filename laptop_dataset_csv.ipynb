```python
# ======================================
# Laptop Data Analysis Project
# ======================================

# Import Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

import warnings
warnings.filterwarnings('ignore')

plt.style.use('ggplot')
sns.set(font_scale=1.1)

print("Libraries Imported Successfully")
```

```python
# Upload Dataset

uploaded = files.upload()

df = pd.read_csv(list(uploaded.keys())[0])

print("Dataset Loaded Successfully")

df.head()
```

```python
print("Shape of Dataset")
print(df.shape)

print("\n")

print("Columns")
print(df.columns)

print("\n")

print("Information")

df.info()

print("\n")

print("Summary Statistics")

display(df.describe(include='all'))

print("\n")

print("Missing Values")

display(df.isnull().sum())
```

```python
# Remove Index Column

if 'Unnamed: 0' in df.columns:
    df.drop(columns=['Unnamed: 0'], inplace=True)

# Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print("Duplicates Removed")

print(df.shape)
```

```python
# Clean Price

df['Price'] = (
    df['Price']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.extract('(\d+)')[0]
)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Clean RAM

df['RAM'] = (
    df['RAM']
    .astype(str)
    .str.extract('(\d+)')[0]
)

df['RAM'] = pd.to_numeric(df['RAM'], errors='coerce')

# Processor Speed

df['Ghz'] = (
    df['Ghz']
    .astype(str)
    .str.extract('([\d.]+)')[0]
)

df['Ghz'] = pd.to_numeric(df['Ghz'], errors='coerce')

# Display Size

df['Display'] = (
    df['Display']
    .astype(str)
    .str.extract('([\d.]+)')[0]
)

df['Display'] = pd.to_numeric(df['Display'], errors='coerce')

# SSD

df['SSD'] = (
    df['SSD']
    .astype(str)
    .str.extract('(\d+)')[0]
)

df['SSD'] = pd.to_numeric(df['SSD'], errors='coerce')

# HDD

df['HDD'] = (
    df['HDD']
    .astype(str)
    .str.extract('(\d+)')[0]
)

df['HDD'] = pd.to_numeric(df['HDD'], errors='coerce')

# Battery

df['Battery_Life'] = (
    df['Battery_Life']
    .astype(str)
    .str.extract('(\d+)')[0]
)

df['Battery_Life'] = pd.to_numeric(df['Battery_Life'], errors='coerce')

print("Cleaning Completed")

df.head()
```

```python
print(df.isnull().sum())

# Fill Numerical Missing Values

numeric = df.select_dtypes(include=np.number).columns

for i in numeric:
    df[i].fillna(df[i].median(), inplace=True)

print("Missing Values Filled")
```

```python
print("Total Laptops :", len(df))

print("Brands :", df['Brand'].nunique())

print("Processor Brands :", df['Processor_Brand'].nunique())

print("GPU Brands :", df['GPU_Brand'].nunique())

print("RAM Types :", df['RAM_TYPE'].nunique())

print("Display Types :", df['Display_type'].nunique())
```
